### What did you like best about the 2016 Aon Staff Wellness Program?

```python
# Step 1: Extract Respondent ID and the raw response for "liked best" question
liked_best_df = df[['Respondent ID',
                    'What did you like best about the 2016 Aon Staff Wellness Program?']].copy()

# Step 2: Drop rows with no response
liked_best_df = liked_best_df.dropna(subset=[
    'What did you like best about the 2016 Aon Staff Wellness Program?'
])

# Step 3: Keep only valid Respondent IDs
liked_best_df = liked_best_df[liked_best_df['Respondent ID'].notna()]

# Step 4: Rename columns for simplicity
liked_best_df.columns = ['respondent_id', 'liked_best']

# Step 5: Preview the first 5 responses
liked_best_df.head(5)
```

```python
# Step 6: Normalize the text for easier keyword matching
liked_best_df['liked_best_clean'] = (
    liked_best_df['liked_best']
    .str.lower()  # lowercase all text
    .str.replace(r'[^\w\s]', '', regex=True)  # remove punctuation
    .str.strip()  # remove extra spaces
)

liked_best_df.head(5)  # Preview normalized text
```

```python
# Step 7: Define themes and associated keywords
themes = {
    'Fitness / Physical Activities': ['gym', 'zumba', 'exercise', 'dancing', 'fitness', 'wellness tests', 'checkups'],
    'Medical / Health Checkups': ['medical', 'checkup', 'hiv', 'condom', 'bmi', 'tests', 'body check'],
    'Learning / Education / Talks': ['talk', 'education', 'workshop', 'professional', 'seminar', 'session'],
    'Staff Engagement / Team Building': ['staff engagement', 'team building', 'participation', 'organisers'],
    'Communication / Awareness / Planning': ['communication', 'planning', 'awareness', 'newsletter', 'method', 'demonstration'],
    'Facilities / Program Logistics': ['launch', 'branches', 'convenience', 'program', 'devolved', 'office']
}
```

```python
# Step 8: Function to assign themes to each response
def assign_themes(text):
    matched_themes = []
    for theme, keywords in themes.items():
        for kw in keywords:
            if kw in text:
                matched_themes.append(theme)
                break  # stop after first keyword match per theme
    if not matched_themes:
        matched_themes.append('Miscellaneous / N/A')
    return matched_themes

# Apply the function to cleaned responses
liked_best_df['themes'] = liked_best_df['liked_best_clean'].apply(assign_themes)
liked_best_df.head(5)  # Preview theme assignment
```

```python
# Step 9: Explode the list of themes for counting multiple themes per response
liked_best_exploded = liked_best_df.explode('themes')
liked_best_exploded.head(5)
```

```python
# Step 10: Count responses per theme and calculate percentage
theme_summary = (
    liked_best_exploded['themes']
    .value_counts()
    .reset_index()
)
theme_summary.columns = ['theme', 'count']

# Add percentage column
theme_summary['percentage'] = (theme_summary['count'] / theme_summary['count'].sum() * 100).round(2)

# Display the summary
theme_summary
```

```python
import matplotlib.pyplot as plt

# Step 12: Prepare labels with theme + percentage for the legend
legend_labels = [
    f"{theme} ({pct}%)" for theme, pct in zip(theme_summary['theme'], theme_summary['percentage'])
]

# Step 13: Plot pie chart
plt.figure(figsize=(5,5))  # Figure size

plt.pie(
    theme_summary['percentage'],  # Percentages for slices
    labels=None,                  # No labels on slices
    autopct='%1.1f%%',            # Show percentages on slices
    startangle=90,                # Start at top
    colors=plt.cm.tab20.colors    # Distinct colors
)

# Add title
plt.title('Themes: What Respondents Liked Best About 2016 Wellness Program', fontsize=14)

# Add legend with percentages
plt.legend(
    legend_labels, 
    title="Themes", 
    bbox_to_anchor=(1.05, 1), 
    loc='upper left'
)

# Make pie circular
plt.axis('equal')
```

```python
# Step 6: Normalize text for keyword matching
suggestions_df['suggestion_clean'] = (
    suggestions_df['suggestion']
    .str.lower()                                  # Convert to lowercase
    .str.replace(r'[^\w\s]', '', regex=True)      # Remove punctuation
    .str.strip()                                  # Remove extra spaces
)

suggestions_df.head()
```

```python
import streamlit as st
import pandas as pd
import plotly.express as px
import ast

# =========================
# PAGE CONFIG
# =========================
st.set_page_config(
    page_title="Wellness & Benefits Dashboard",
    page_icon="📊",
    layout="wide"
)

# =========================
# LOAD DATA FUNCTION
# =========================
@st.cache_data
def load_data(client, year):
    if client == "ABSA" and year == 2022:
        df = pd.read_csv("master_survey_data.csv")
        df.columns = df.columns.str.strip().str.lower()
    elif client == "AON" and year == 2020:
        df = pd.read_csv("aon_2020_master.csv")
        df.columns = df.columns.str.strip().str.lower()
    else:
        df = pd.DataFrame()
    return df

# =========================
# SIDEBAR FILTERS
# =========================
st.sidebar.title("🎛 Dashboard Filters")
year = st.sidebar.selectbox("📅 Survey Year", [2020, 2021, 2022, 2023, 2024, 2025])
client = st.sidebar.selectbox("🏦 Client", ["AON", "ABSA"])

df = load_data(client, year)
if df.empty:
    st.warning("No data available for this client/year combination.")
    st.stop()

# -------------------------
# Gender Filter
# -------------------------
gender_filter = st.sidebar.multiselect(
    "Select Gender",
    options=df["gender"].dropna().unique(),
    default=df["gender"].dropna().unique()
)

# -------------------------
# Dynamic Age Column Handling
# -------------------------
age_col = None
for col in ["age_band", "age_group", "age"]:
    if col in df.columns:
        age_col = col
        break

age_filter = None
if age_col:
    age_filter = st.sidebar.multiselect(
        "Select Age Band",
        options=df[age_col].dropna().unique(),
        default=df[age_col].dropna().unique()
    )

# -------------------------
# Filter Data Based on Sidebar
# -------------------------
filtered_df = df[df["gender"].isin(gender_filter)]
if age_filter is not None:
    filtered_df = filtered_df[filtered_df[age_col].isin(age_filter)]

# -------------------------
# Client-Specific Filters
# -------------------------
if client == "ABSA":
    branch_filter = st.sidebar.multiselect(
        "Select Branch",
        options=df["branch"].dropna().unique(),
        default=df["branch"].dropna().unique()
    )
    function_filter = st.sidebar.multiselect(
        "Select Function / Role",
        options=df["function"].dropna().unique(),
        default=df["function"].dropna().unique()
    )
    filtered_df = filtered_df[
        (filtered_df["branch"].isin(branch_filter)) &
        (filtered_df["function"].isin(function_filter))
    ]
elif client == "AON":
    county_filter = st.sidebar.multiselect(
        "Select County",
        options=df["county"].dropna().unique(),
        default=df["county"].dropna().unique()
    )
    dependents_filter = st.sidebar.multiselect(
        "Select Household Dependents Category",
        options=df["dependents_category"].dropna().unique(),
        default=df["dependents_category"].dropna().unique()
    )
    filtered_df = filtered_df[
        (filtered_df["county"].isin(county_filter)) &
        (filtered_df["dependents_category"].isin(dependents_filter))
    ]

# =========================
# TITLE
# =========================
st.title(f"📊 {client} {year} Employee Wellness & Benefits Dashboard")
st.caption("Interactive overview of employee wellness, coverage, and benefits insights")

# =========================
# KPIs / METRICS
# =========================
kpi_cols = st.columns(3)

if client == "ABSA":
    total_responses = len(filtered_df)
    aware_pct = filtered_df["awareness_of_wellness_initiative"].value_counts(normalize=True).get("Yes", 0) * 100
    eap_pct = filtered_df["link_to_eap"].value_counts(normalize=True).get("Yes", 0) * 100

    kpi_cols[0].metric("📋 Total Survey Responses", total_responses)
    kpi_cols[1].metric("📢 Awareness of Wellness Initiatives", f"{aware_pct:.1f}%")
    kpi_cols[2].metric("🤝 Linked to Employee Assistance Program (EAP)", f"{eap_pct:.1f}%")

elif client == "AON":
    total_responses = len(filtered_df)
    coverage_gap = filtered_df["all_direct_covered"].value_counts(normalize=True).get("No", 0) * 100
    willing_pct = filtered_df["willingness_to_pay"].isin(["Willing", "Very willing"]).sum() / total_responses * 100

    kpi_cols[0].metric("📋 Total Survey Responses", total_responses)
    kpi_cols[1].metric("❌ Coverage Gap (Direct Dependents)", f"{coverage_gap:.1f}%")
    kpi_cols[2].metric("💰 Willingness to Pay for Additional Coverage", f"{willing_pct:.1f}%")

st.markdown("---")

# =========================
# DEMOGRAPHICS ROW
# =========================
c1, c2, c3 = st.columns(3)

# Gender Pie Chart
with c1:
    fig_gender = px.pie(
        filtered_df,
        names="gender",
        hole=0.45,
        title="Gender Distribution of Respondents",
        color_discrete_sequence=px.colors.qualitative.Set2
    )
    st.plotly_chart(fig_gender, use_container_width=True)

# Age Distribution
with c2:
    if age_col:
        filtered_df["age_band"] = filtered_df[age_col]
        age_counts = filtered_df["age_band"].value_counts().reset_index()
        age_counts.columns = ["Age Band", "Count"]
        fig_age = px.bar(
            age_counts,
            x="Age Band",
            y="Count",
            title="Age Distribution of Respondents",
            color_discrete_sequence=px.colors.qualitative.Bold
        )
        st.plotly_chart(fig_age, use_container_width=True)
    else:
        st.warning("No age/age_band/age_group column found in this dataset.")

# Awareness / Coverage Pie
with c3:
    if client == "ABSA":
        fig_awareness = px.pie(
            filtered_df,
            names="awareness_of_wellness_initiative",
            hole=0.45,
            title="Awareness of Wellness Initiatives",
            color_discrete_sequence=px.colors.qualitative.Pastel
        )
        st.plotly_chart(fig_awareness, use_container_width=True)
    elif client == "AON":
        fig_coverage = px.pie(
            filtered_df,
            names="all_direct_covered",
            hole=0.45,
            title="Direct Dependents Coverage Status",
            color_discrete_sequence=px.colors.qualitative.Pastel
        )
        st.plotly_chart(fig_coverage, use_container_width=True)

st.markdown("---")

# =========================
# ABSA 2022 FULL SECTIONS
# =========================
if client == "ABSA" and year == 2022:

    # Survey Coverage
    st.markdown("## 📌 Survey Coverage & Context")
    c1, c2, c3, c4 = st.columns(4)
    c1.metric("👥 Total Respondents", len(filtered_df))
    c2.metric("🏢 Branches Covered", filtered_df["branch"].nunique())
    c3.metric("🧑‍💼 Functions Covered", filtered_df["function"].nunique())
    aware_pct = filtered_df["awareness_of_wellness_initiative"].value_counts(normalize=True).get("Yes", 0) * 100
    c4.metric("📢 Awareness of Wellness", f"{aware_pct:.1f}%")
    st.markdown("---")

    # Respondent Profile
    st.markdown("## 👥 Respondent Profile")
    d1, d2 = st.columns(2)
    with d1:
        fig_gender = px.pie(filtered_df, names="gender", hole=0.5, color_discrete_sequence=px.colors.qualitative.Set2)
        st.plotly_chart(fig_gender, use_container_width=True)
    with d2:
        age_counts = filtered_df["age_group"].value_counts().reset_index()
        age_counts.columns = ["Age Group", "Count"]
        fig_age = px.bar(age_counts, x="Age Group", y="Count", text="Count", color_discrete_sequence=px.colors.qualitative.Bold)
        st.plotly_chart(fig_age, use_container_width=True)
    st.markdown("---")

    # Post-COVID Challenges
    st.markdown("## 🌪 Post-COVID Challenges Faced by Employees")
    post_covid = filtered_df["post_covid_issues"].dropna().value_counts().reset_index()
    post_covid.columns = ["Challenge", "Count"]
    fig_pc = px.bar(post_covid, x="Count", y="Challenge", orientation="h", color="Count", color_continuous_scale=px.colors.sequential.Viridis)
    st.plotly_chart(fig_pc, use_container_width=True)

    # Financial Stressors
    st.markdown("## 💸 Financial Stressors Affecting Wellbeing")
    financial = filtered_df["financial_issues"].dropna().value_counts().reset_index()
    financial.columns = ["Financial Issue", "Count"]
    if not financial.empty:
        fig_fin = px.bar(financial, x="Count", y="Financial Issue", orientation="h", color="Count", color_continuous_scale=px.colors.sequential.Plasma)
        st.plotly_chart(fig_fin, use_container_width=True)
    else:
        st.info("No financial stressor data available for the selected filters.")

    # Requested Wellness Support
    st.markdown("## 🧩 Requested Areas for Wellness Program Inclusion")
    needs = filtered_df["wellness_program_needs"].dropna().value_counts().reset_index()
    needs.columns = ["Requested Support Area", "Count"]
    fig_needs = px.bar(needs, x="Count", y="Requested Support Area", orientation="h", color="Count", color_continuous_scale=px.colors.sequential.Magma)
    st.plotly_chart(fig_needs, use_container_width=True)

    # Workplace Investment Priorities
    st.markdown("## 🏗 Workplace Investment Priorities for Productivity")
    priority_cols = [
        "effective communication skills among colleagues",
        "interpersonal relationships with colleagues",
        "employee - manager working relationships",
        "embracing teamwork among colleagues",
        "developing social support structures among team members"
    ]
    priority_df = filtered_df[priority_cols].apply(pd.Series.value_counts).fillna(0).T.reset_index().rename(columns={"index":"Area"})
    fig_env = px.bar(priority_df, x="Highly Important", y="Area", orientation="h", color="Highly Important", color_continuous_scale=px.colors.sequential.Cividis)
    st.plotly_chart(fig_env, use_container_width=True)

    # Disclosure & EAP
    st.markdown("## 🤝 Disclosure & Support Escalation")
    e1, e2 = st.columns(2)
    with e1:
        fig_issue = px.pie(filtered_df, names="has_health_wellbeing_issue", hole=0.5)
        st.plotly_chart(fig_issue, use_container_width=True)
    with e2:
        fig_eap = px.pie(filtered_df, names="link_to_eap", hole=0.5)
        st.plotly_chart(fig_eap, use_container_width=True)

    # Employee-Voiced Support Requests
    st.markdown("## 🗣 Employee-Voiced Support Requests")
    filtered_df["themes"] = filtered_df["themes"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else [])
    all_requests = [t for sub in filtered_df["themes"] for t in sub]
    req_counts = pd.Series(all_requests).value_counts().reset_index()
    req_counts.columns = ["Support Request Category", "Mentions"]
    fig_req = px.bar(req_counts, x="Mentions", y="Support Request Category", orientation="h", color="Mentions", color_continuous_scale=px.colors.sequential.Aggrnyl)
    st.plotly_chart(fig_req, use_container_width=True)

# =========================
# AON 2020 Sections
# =========================
elif client == "AON" and year == 2020:
    c1, c2 = st.columns(2)
    with c1:
        dependents_counts = filtered_df["dependents_category"].value_counts().reset_index()
        dependents_counts.columns = ["Household Dependents Category", "Count"]
        fig_dep = px.bar(dependents_counts, x="Household Dependents Category", y="Count", color="Count", color_continuous_scale=px.colors.sequential.Plasma)
        st.plotly_chart(fig_dep, use_container_width=True)
    with c2:
        county_counts = filtered_df["county"].value_counts().nlargest(10).reset_index()
        county_counts.columns = ["County", "Responses"]
        fig_county = px.bar(county_counts, y="County", x="Responses", orientation="h", color="Responses", color_continuous_scale=px.colors.sequential.Viridis)
        st.plotly_chart(fig_county, use_container_width=True)
    st.markdown("### 💰 Willingness to Pay for Additional Dependent Coverage")
    if "willingness_to_pay" in filtered_df.columns:
        willingness_counts = filtered_df["willingness_to_pay"].value_counts().reset_index()
        willingness_counts.columns = ["Response", "Count"]
        fig_willing = px.pie(willingness_counts, names="Response", values="Count", hole=0.4, color_discrete_sequence=px.colors.qualitative.Set3)
        st.plotly_chart(fig_willing, use_container_width=True)
    if "amount_selected" in filtered_df.columns:
        amount_counts = filtered_df["amount_selected"].value_counts().reset_index()
        amount_counts.columns = ["Preferred Annual Amount per Dependent", "Count"]
        fig_amount = px.bar(amount_counts, x="Preferred Annual Amount per Dependent", y="Count", color="Count", color_continuous_scale=px.colors.sequential.Magma)
        st.plotly_chart(fig_amount, use_container_width=True)

# =========================
# INSIGHTS
# =========================
st.markdown("## 📌 Key Insights")
if client == "ABSA":
    st.write("""
    • Majority of respondents fall within the 25–44 age brackets, indicating a working-age dominant workforce.  
    • Awareness of wellness initiatives is generally high, but linkage to EAP remains lower — suggesting a utilization gap.  
    • Work stressors, mental health concerns, and flexibility needs emerge strongly across wellbeing themes.  
    • Branch and function-level differences highlight opportunities for targeted interventions.
    """)
elif client == "AON":
    st.write("""
    • Most respondents have 1–4 dependents; a significant portion have more than 4 dependents.  
    • Coverage for all direct dependents is not universal — highlighting gaps in employee benefits.  
    • Willingness to pay is mixed, with some employees neutral or unwilling.  
    • Preferred annual amounts vary widely, indicating diverse affordability or expectations.
    """)

st.caption("Wellness & Benefits Dashboard | Streamlit ✅")
```
