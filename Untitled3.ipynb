```python
import pandas as pd
import glob
import os
from docx import Document
from docx.shared import Inches

# Folder path
folder_path = r"C:\Users\Administrator\Desktop\Minet"

# Get all CSV files
csv_files = glob.glob(os.path.join(folder_path, "*.csv"))

# Create a new Word document
doc = Document()
doc.add_heading("Survey Data Previews", level=0)

# Loop through CSVs and add preview tables
for file in csv_files:
    filename = os.path.basename(file)
    try:
        df = pd.read_csv(file)
        df.columns = df.columns.str.strip()
        doc.add_heading(filename, level=1)
        
        # Add first 5 rows as a table
        preview = df.head()
        table = doc.add_table(rows=preview.shape[0]+1, cols=preview.shape[1])
        table.style = 'Light List Accent 1'
        
        # Add column headers
        for j, col in enumerate(preview.columns):
            table.cell(0, j).text = str(col)
        
        # Add row values
        for i in range(preview.shape[0]):
            for j in range(preview.shape[1]):
                table.cell(i+1, j).text = str(preview.iloc[i, j])
        
    except Exception as e:
        doc.add_paragraph(f"Error loading {filename}: {e}")

# Save Word document
output_file = os.path.join(folder_path, "Survey_Previews.docx")
doc.save(output_file)
print(f"Word document with previews saved to {output_file}")
```
